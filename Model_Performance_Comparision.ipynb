```python
from tensorflow.keras.datasets import imdb
(train_data, train_labels), (test_data, test_labels) = imdb.load_data(num_words=10000)
print(train_data)
```

```python
import numpy as np
def vectoroze_sequence(sequences, dimensions = 10000):
  results = np.zeros((len(sequences), dimensions))
  for i, sequence in enumerate(sequences):
    results[i, sequence] = 1
  return results
x_train = vectoroze_sequence(train_data)
x_test = vectoroze_sequence(test_data)
```

```python
from tensorflow import keras
model = keras.Sequential([
   keras.layers.Dense(16, activation="relu"),
   keras.layers.Dense(16, activation="relu"),
   keras.layers.Dense(1, activation="sigmoid")
])
```

```python
model.compile(optimizer="rmsprop", loss="binary_crossentropy", metrics=["accuracy"])
```

```python
history_original = model.fit(x_train, train_labels, epochs=20, batch_size=512, validation_split=0.2)
```

```python
from tensorflow import keras
model = keras.Sequential([
   keras.layers.Dense(4, activation="relu"),
   keras.layers.Dense(4, activation="relu"),
   keras.layers.Dense(1, activation="sigmoid")
])
```

```python
model.compile(optimizer="rmsprop", loss="binary_crossentropy", metrics=["accuracy"])
```

```python
history_smaller_model = model.fit(x_train, train_labels, epochs=20, batch_size=512, validation_split=0.2)
```

```python
from tensorflow import keras
model = keras.Sequential([
   keras.layers.Dense(512, activation="relu"),
   keras.layers.Dense(512, activation="relu"),
   keras.layers.Dense(1, activation="sigmoid")
])
```

```python
model.compile(optimizer="rmsprop", loss="binary_crossentropy", metrics=["accuracy"])
history_large_model = model.fit(x_train, train_labels, epochs=20, batch_size=512, validation_split=0.2)
```

```python
import matplotlib.pyplot as plt
history_org_dict = history_original.history
loss_org_values = history_org_dict["loss"]
history_small_dict = history_smaller_model.history
loss_small_values = history_small_dict["loss"]
history_large_dict = history_large_model.history
loss_large_values = history_large_dict["loss"]
epochs = range(1, len(loss_org_values) + 1)
plt.plot(epochs, loss_org_values, "r", label="Original Model Loss")
plt.plot(epochs, loss_small_values, "b", label="Small Model Loss")
plt.plot(epochs, loss_large_values, "g", label="Large Model Loss")
plt.title("Original vs Small vs Larger Model Loss")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()
plt.show()
```

```python
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras import regularizers
model = keras.Sequential([
  layers.Dense(16, kernel_regularizer = regularizers.l2(0.002), activation="relu"),
  layers.Dense(16, kernel_regularizer = regularizers.l2(0.002), activation="relu"),
  layers.Dense(1, activation="sigmoid")
])
```

```python
model.compile(optimizer="rmsprop", loss="binary_crossentropy", metrics=["accuracy"])
history_l2_model = model.fit(x_train, train_labels, epochs=20, batch_size=512, validation_split=0.2)
```

```python
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras import regularizers
model = keras.Sequential([
  layers.Dense(16, activation="relu"),
  layers.Dropout(0.5),
  layers.Dense(16, activation="relu"),
  layers.Dropout(0.5),
  layers.Dense(1, activation="sigmoid")
])
```

```python
model.compile(optimizer="rmsprop", loss="binary_crossentropy", metrics=["accuracy"])
history_dropout_model = model.fit(x_train, train_labels, epochs=20, batch_size=512, validation_split=0.2)
```

```python
import matplotlib.pyplot as plt
history_org_dict = history_original.history
loss_org_values = history_org_dict["loss"]
history_l2_dict = history_l2_model.history
loss_l2_values = history_l2_dict["loss"]
history_l2_dict = history_dropout_model.history
loss_dropout_values = history_l2_dict["loss"]
epochs = range(1, len(loss_org_values) + 1)
plt.plot(epochs, loss_org_values, "r", label="Original Model Loss")
plt.plot(epochs, loss_l2_values, "b", label="L2 Regularized Model Loss")
plt.plot(epochs, loss_dropout_values, "g", label="L2 Regularized Model Loss")
plt.title("Original vs L2 Regularized Vs Dropout Loss")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()
plt.show()
```
